```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("uae_used_cars_10k.csv")

# Display basic information
print("Dataset shape:", df.shape)
display(df.head())
```

## 1. Regression Error Analysis

### 1.1 Residual Analysis

For the regression task, the target variable is **Price**, since the goal is to predict the price of a used car based on its characteristics.

A **5-fold cross-validation** approach will be used. The dataset contains approximately 10,000 observations, so k=5 provides a good balance between computational efficiency and reliable performance estimation. It also limits the variance associated with using a single train-test split.

The residual for each observation is defined as:

**Residual = Actual Price - Predicted Price**


```python
# Select features and target for regression
reg_features = [
    "Year",
    "Mileage",
    "Cylinders",
    "Make",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Location"
]

X = df[reg_features].copy()
y = df["Price"].copy()

# Remove rows with missing values
reg_data = pd.concat([X, y], axis=1).dropna()

X = reg_data[reg_features]
y = reg_data["Price"]

print("Regression dataset shape:", X.shape)
print("\nTarget variable: Price")
print("\nFeatures:")
print(X.columns.tolist())

```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Convert numerical columns to numeric
for col in ["Year", "Mileage", "Cylinders", "Price"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Select regression features
reg_features = [
    "Year",
    "Mileage",
    "Cylinders",
    "Make",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Location"
]

# Keep only valid rows
reg_data = df[reg_features + ["Price"]].dropna().copy()

X = reg_data[reg_features]
y = reg_data["Price"]

# Separate numerical and categorical features
numeric_features = ["Year", "Mileage", "Cylinders"]

categorical_features = [
    "Make",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Location"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Linear Regression pipeline
regression_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

# 5-fold cross-validation
kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Out-of-fold predictions
y_pred = cross_val_predict(
    regression_model,
    X,
    y,
    cv=kf
)

# Calculate residuals
residuals = y.to_numpy() - y_pred

# Performance metrics
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
r2 = r2_score(y, y_pred)

print("Regression dataset shape:", X.shape)
print("\n5-Fold Cross-Validation Results")
print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.4f}")
print(f"Mean Residual: {residuals.mean():.2f}")
```

```python
# Residuals vs predicted values
plt.figure(figsize=(10, 6))

sns.scatterplot(
    x=y_pred,
    y=residuals,
    alpha=0.4
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Residuals vs Predicted Values")
plt.xlabel("Predicted Price")
plt.ylabel("Residual")

plt.show()


# Distribution of residuals
plt.figure(figsize=(10, 6))

sns.histplot(
    residuals,
    bins=50,
    kde=True
)

plt.axvline(
    x=0,
    linestyle="--"
)

plt.title("Distribution of Residuals")
plt.xlabel("Residual")
plt.ylabel("Frequency")

plt.show()
```

### Residual Analysis

The residuals are generally concentrated around zero, but the distribution contains several large positive and negative residuals. The mean residual is approximately 276, which is very small relative to the scale of vehicle prices, suggesting that the model does not have a strong overall directional bias.

The residual plot shows that the variance of the errors is not constant. As predicted prices increase, the spread of the residuals becomes larger, indicating evidence of heteroscedasticity.

Several extreme positive residuals are visible, meaning that for some expensive vehicles the model substantially underestimates the actual price. The residual distribution is therefore not symmetric and is strongly affected by extreme observations.

The higher-priced vehicle segment appears to be more difficult for the linear regression model. This may be caused by luxury or unusual vehicles whose prices depend on characteristics that are not fully captured by the available features, as well as the limitations of a linear model in representing complex relationships.

```python
# 1.2 - Error as a Function of Features

# Create a dataframe for error analysis
error_df = X.copy()

error_df["Actual_Price"] = y.values
error_df["Predicted_Price"] = y_pred
error_df["Residual"] = residuals
error_df["Absolute_Error"] = np.abs(residuals)

# Analyze error as a function of Year and Mileage
features_to_analyze = ["Year", "Mileage"]

for feature in features_to_analyze:

    plt.figure(figsize=(10, 5))

    sns.scatterplot(
        data=error_df,
        x=feature,
        y="Absolute_Error",
        alpha=0.4
    )

    plt.title(f"Absolute Error vs {feature}")
    plt.xlabel(feature)
    plt.ylabel("Absolute Error")

    plt.show()
```

### Error as a Function of Features

For Year, most observations have relatively small absolute errors across all manufacturing years. However, several extreme errors are present in specific years, indicating that some vehicles are considerably more difficult for the model to predict.

For Mileage, most prediction errors remain relatively small across the mileage range. There is no clear linear relationship between mileage and absolute error, although several large outliers are visible.

Overall, the results suggest that Year and Mileage alone do not fully explain the largest prediction errors. The extreme errors are likely related to additional characteristics of particular vehicles, such as manufacturer, model, body type, or other factors not fully captured by the linear regression model

## 2. Regression Models

In this section, three regression models are evaluated and compared using the same feature set and 5-fold cross-validation:

1. Linear Regression
2. Decision Tree Regressor
3. Random Forest Regressor

The models are evaluated using MAE, MSE, RMSE, and R² to ensure a fair comparison.

```python
# 2. Regression Models

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Identify numerical and categorical features
numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

# Preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Models and hyperparameters
models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=10,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=10,
        random_state=42,
        n_jobs=-1
    )
}

# 5-fold cross-validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    predictions = cross_val_predict(
        pipeline,
        X,
        y,
        cv=kf,
        n_jobs=-1
    )

    mae = mean_absolute_error(y, predictions)
    mse = mean_squared_error(y, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    })

# Results table
results_df = pd.DataFrame(results)

results_df
```

### 2.1 Discussion and Model Comparison

The three regression models were evaluated using the same feature set and 5-fold cross-validation to ensure a fair comparison.

The Random Forest Regressor achieved the best overall predictive performance. It produced the lowest MAE and RMSE and the highest R² value (approximately 0.495). The Decision Tree Regressor performed better than Linear Regression, but worse than Random Forest. Linear Regression showed the weakest overall performance.

The weaker performance of Linear Regression suggests that the relationship between the vehicle characteristics and price is not purely linear. In addition, the residual analysis showed heteroscedasticity, strong skewness, and extreme outliers, which violate important assumptions associated with linear regression.

The Decision Tree can capture non-linear relationships and interactions between features, giving it an advantage over Linear Regression. However, a single decision tree can have high variance and may be sensitive to changes in the training data and to outliers.

Random Forest reduces the variance of individual decision trees by combining predictions from multiple trees. This provides a better bias-variance trade-off and makes the model more robust to noise and complex non-linear relationships.

The previous error analysis also showed that rare and expensive luxury vehicles generate very large prediction errors. All models may be affected by these extreme observations, although tree-based ensemble models are generally better able to model complex patterns than a simple linear model.

There is also a trade-off between interpretability and predictive performance. Linear Regression is relatively easy to interpret, while Random Forest is more complex. However, in this dataset, the improvement in predictive performance makes Random Forest the preferred model for future analysis.

Overall, the modeling stage demonstrates that model choice has a substantial effect on prediction accuracy. The results suggest that vehicle prices depend on complex and non-linear relationships between the available features, and that an ensemble tree-based model is more appropriate for this dataset.


```python
# ============================================================
# 3. Classification Error Analysis
# 3.1 Confusion Matrix Analysis
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ------------------------------------------------------------
# Define the classification problem
# 1 = High-price vehicle
# 0 = Lower-price vehicle
# ------------------------------------------------------------

price_threshold = df["Price"].median()

df_class = df.copy()
df_class["High_Price"] = (df_class["Price"] > price_threshold).astype(int)

print("Price threshold (median):", price_threshold)
print("\nClass distribution:")
print(df_class["High_Price"].value_counts())

# ------------------------------------------------------------
# Use the same feature set as the regression analysis
# ------------------------------------------------------------

classification_features = [
    "Year",
    "Mileage",
    "Cylinders",
    "Make",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Location"
]

X_class = df_class[classification_features].copy()
y_class = df_class["High_Price"].copy()

# Identify numerical and categorical features
numeric_features = ["Year", "Mileage", "Cylinders"]

categorical_features = [
    "Make",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Location"
]

# ------------------------------------------------------------
# Preprocessing
# ------------------------------------------------------------

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_class = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# ------------------------------------------------------------
# Classification model
# ------------------------------------------------------------

classification_model = Pipeline(steps=[
    ("preprocessor", preprocessor_class),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# ------------------------------------------------------------
# 5-Fold Stratified Cross-Validation
# ------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

y_class_pred = cross_val_predict(
    classification_model,
    X_class,
    y_class,
    cv=skf,
    method="predict"
)

# ------------------------------------------------------------
# Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(y_class, y_class_pred)

tn, fp, fn, tp = cm.ravel()

print("\nConfusion Matrix:")
print(cm)

print("\nTN =", tn)
print("FP =", fp)
print("FN =", fn)
print("TP =", tp)

# ------------------------------------------------------------
# Performance metrics
# ------------------------------------------------------------

print("\nClassification Metrics")
print("----------------------")
print("Accuracy :", round(accuracy_score(y_class, y_class_pred), 4))
print("Precision:", round(precision_score(y_class, y_class_pred), 4))
print("Recall   :", round(recall_score(y_class, y_class_pred), 4))
print("F1-score :", round(f1_score(y_class, y_class_pred), 4))

print("\nClassification Report:")
print(classification_report(
    y_class,
    y_class_pred,
    target_names=["Lower Price", "High Price"]
))

# ------------------------------------------------------------
# Visualize Confusion Matrix
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Lower Price", "High Price"],
    yticklabels=["Lower Price", "High Price"]
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("3.1 Confusion Matrix - Logistic Regression")

plt.show()
```


### 3.1 Confusion Matrix Analysis – Interpretation

The Logistic Regression classifier achieved an accuracy of approximately 73.5% using 5-fold stratified cross-validation.

The confusion matrix contains 3692 true negatives, 3654 true positives, 1308 false positives, and 1346 false negatives.

False positives correspond to lower-priced vehicles that were incorrectly classified as high-priced vehicles, while false negatives correspond to high-priced vehicles that were incorrectly classified as lower-priced vehicles.

The numbers of false positives and false negatives are relatively similar, indicating that the model does not show a strong bias toward one type of classification error.

In this context, false negatives may be considered slightly more important because they represent expensive vehicles that the model incorrectly identifies as lower-priced vehicles. However, the relative importance of FP and FN depends on the practical application of the model.

The results indicate that vehicle characteristics such as year, mileage, make, body type, transmission, fuel type, cylinders, and location provide useful information for distinguishing between the two price classes, but they do not completely explain vehicle price. Additional factors such as vehicle condition, trim level, optional equipment, and market-specific characteristics could improve classification performance

```python
# ============================================================
# 3.2 Probability-Based Analysis
# ============================================================

from sklearn.model_selection import cross_val_predict

# Get out-of-fold predicted probabilities using the same
# Logistic Regression model and 5-fold cross-validation
y_prob = cross_val_predict(
    classification_model,
    X_class,
    y_class,
    cv=skf,
    method="predict_proba"
)[:, 1]

# Convert probabilities to class predictions
y_pred_class = (y_prob >= 0.5).astype(int)

# Identify correct and incorrect predictions
correct_mask = y_pred_class == y_class.to_numpy()
incorrect_mask = ~correct_mask

correct_probs = y_prob[correct_mask]
incorrect_probs = y_prob[incorrect_mask]

# Confidence of each prediction
confidence = np.maximum(y_prob, 1 - y_prob)

print("Number of correct predictions:", correct_mask.sum())
print("Number of incorrect predictions:", incorrect_mask.sum())

print("\nAverage confidence:")
print(f"Correct predictions:   {confidence[correct_mask].mean():.4f}")
print(f"Incorrect predictions: {confidence[incorrect_mask].mean():.4f}")

# High-confidence errors
high_conf_error = incorrect_mask & (confidence >= 0.90)

print("\nHigh-confidence errors (confidence >= 0.90):",
      high_conf_error.sum())

# ------------------------------------------------------------
# Plot predicted probabilities
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(
    correct_probs,
    bins=30,
    alpha=0.6,
    label="Correct predictions"
)

plt.hist(
    incorrect_probs,
    bins=30,
    alpha=0.6,
    label="Incorrect predictions"
)

plt.axvline(
    0.5,
    linestyle="--",
    label="Decision threshold = 0.5"
)

plt.xlabel("Predicted Probability of High Price")
plt.ylabel("Frequency")
plt.title("3.2 Predicted Probabilities: Correct vs Incorrect Predictions")
plt.legend()

plt.show()
```

### 3.2 Probability-Based Analysis – Interpretation

Correct predictions have a higher average confidence (0.7661) than incorrect predictions (0.6493).

Most incorrect predictions occur at intermediate predicted probabilities, particularly around the 0.5 decision threshold. This indicates that the model makes more errors when it is uncertain about the predicted class.

Only 34 incorrect predictions were made with confidence greater than or equal to 0.90. Therefore, high-confidence errors are relatively rare.

Overall, the probability analysis suggests that the model's confidence is generally meaningful: predictions with higher confidence are more likely to be correct, while uncertain predictions are more prone to classification errors.

```python
# ============================================================
# 3.3 Error as a Function of Features
# ============================================================

# Create a dataframe for classification error analysis
classification_error_df = X_class.copy()

classification_error_df["Actual_Class"] = y_class.to_numpy()
classification_error_df["Predicted_Class"] = y_pred_class

classification_error_df["Prediction_Status"] = np.where(
    classification_error_df["Actual_Class"] ==
    classification_error_df["Predicted_Class"],
    "Correct",
    "Incorrect"
)

# ------------------------------------------------------------
# Compare Year distribution for correct vs incorrect predictions
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=classification_error_df,
    x="Prediction_Status",
    y="Year"
)

plt.title("3.3 Manufacturing Year: Correct vs Incorrect Predictions")
plt.xlabel("Prediction Status")
plt.ylabel("Year")

plt.show()


# ------------------------------------------------------------
# Compare Mileage distribution for correct vs incorrect predictions
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=classification_error_df,
    x="Prediction_Status",
    y="Mileage"
)

plt.title("3.3 Mileage: Correct vs Incorrect Predictions")
plt.xlabel("Prediction Status")
plt.ylabel("Mileage")

plt.show()


# ------------------------------------------------------------
# Error rate by manufacturer
# ------------------------------------------------------------

manufacturer_error_rate = (
    classification_error_df
    .groupby("Make")
    .agg(
        Number_of_Cars=("Prediction_Status", "size"),
        Error_Rate=(
            "Prediction_Status",
            lambda x: (x == "Incorrect").mean()
        )
    )
)

# Keep manufacturers with at least 50 observations
manufacturer_error_rate = manufacturer_error_rate[
    manufacturer_error_rate["Number_of_Cars"] >= 50
].sort_values(
    by="Error_Rate",
    ascending=False
)

print("Manufacturers with the highest classification error rates:")
display(manufacturer_error_rate.head(15))
```


### 3.3 Error as a Function of Features – Interpretation

The distributions of manufacturing year and mileage were compared between correct and incorrect predictions.

The distributions show substantial overlap between the two groups. Therefore, neither year nor mileage alone clearly determines whether a vehicle will be classified correctly.

This suggests that classification errors are more likely to result from interactions between multiple vehicle characteristics rather than from a single numerical feature.

The manufacturer-level error analysis can also help identify specific vehicle brands for which the model has greater difficulty, indicating that categorical characteristics may contribute to systematic misclassification

```python
# ============================================================
# 3.4 Threshold Sensitivity Analysis
# ============================================================

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    roc_auc_score,
    roc_curve,
    confusion_matrix
)

# Evaluate thresholds from 0.1 to 0.9
thresholds = np.arange(0.1, 1.0, 0.1)

threshold_results = []

for threshold in thresholds:

    y_threshold_pred = (y_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_class,
        y_threshold_pred
    ).ravel()

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_class,
            y_threshold_pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_class,
            y_threshold_pred,
            zero_division=0
        ),
        "F1": f1_score(
            y_class,
            y_threshold_pred,
            zero_division=0
        ),
        "MCC": matthews_corrcoef(
            y_class,
            y_threshold_pred
        ),
        "FP": fp,
        "FN": fn
    })

threshold_results_df = pd.DataFrame(threshold_results)

display(threshold_results_df)

# ------------------------------------------------------------
# Plot Precision, Recall, F1 and MCC across thresholds
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

for metric in ["Precision", "Recall", "F1", "MCC"]:
    plt.plot(
        threshold_results_df["Threshold"],
        threshold_results_df[metric],
        marker="o",
        label=metric
    )

plt.title("3.4 Performance Across Classification Thresholds")
plt.xlabel("Threshold")
plt.ylabel("Score")
plt.xticks(thresholds)
plt.legend()
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot False Positives and False Negatives across thresholds
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    threshold_results_df["Threshold"],
    threshold_results_df["FP"],
    marker="o",
    label="False Positives"
)

plt.plot(
    threshold_results_df["Threshold"],
    threshold_results_df["FN"],
    marker="o",
    label="False Negatives"
)

plt.title("False Positives and False Negatives Across Thresholds")
plt.xlabel("Threshold")
plt.ylabel("Number of Errors")
plt.xticks(thresholds)
plt.legend()
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# ROC Curve and AUC
# ------------------------------------------------------------

fpr, tpr, roc_thresholds = roc_curve(
    y_class,
    y_prob
)

auc_score = roc_auc_score(
    y_class,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {auc_score:.4f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid(True)

plt.show()

print(f"AUC-ROC: {auc_score:.4f}")

```

### 3.4 Threshold Sensitivity Analysis – Interpretation

The classification threshold has a clear effect on the trade-off between false positives and false negatives.

At low thresholds, the model classifies more vehicles as high-priced. As a result, recall is higher, but the number of false positives increases. As the threshold increases, false positives decrease while false negatives increase.

Around the default threshold of 0.5, false positives and false negatives are relatively balanced. Therefore, a threshold near 0.5 provides a reasonable operating point when both types of errors are considered similarly important.

Extreme thresholds are less stable because they strongly favor one type of error over the other. Intermediate thresholds provide a more balanced trade-off between precision and recall.

The ROC analysis produced an AUC-ROC score of 0.8231. This indicates that the model has good discriminatory ability and performs substantially better than random classification.

Overall, the threshold should be selected according to the practical cost of false positives and false negatives. For the current problem, a threshold around 0.5 provides a reasonable balance
